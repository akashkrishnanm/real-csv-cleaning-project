```python
df["Income"].fillna(df["Income"].mean(), inplace=True)
```

## Renaming Columns

```python
df.rename(columns={
     "MntWines": "Wine_Spending",
    "MntFruits": "Fruit_Spending",
    "NumDealsPurchases": "Deal_Purchases"
}, inplace=True)
```

```python
df.columns
```

```python
df.dtypes
```

```python
df["Dt_Customer"].head()
```

## Datatype Conversion

```python
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True)
```

```python
df["Dt_Customer"].dtype
```

```python
df["Dt_Customer"].dt.year
```

```python
df["Dt_Customer"].dt.month
```

```python
df.sort_values("Dt_Customer")
```

## Exporting Cleaned Dataset

```python
df.to_csv("cleaned_marketing_campaign.csv", index=False)
```

## Conclusion

- Missing values in the Income column were handled using mean imputation.
- Duplicate rows were checked and no duplicates were found.
- Several abbreviated column names were renamed for better readability.
- The customer date column was converted into datetime format for future analysis.
- A cleaned version of the dataset was successfully exported.
